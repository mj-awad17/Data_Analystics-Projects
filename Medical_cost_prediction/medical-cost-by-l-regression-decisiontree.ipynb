    <h4 style="color: white; font-family: 'Algerian', serif;">📋 Dataset Columns:</h4>
    <ul>
        <li><strong>age:</strong> The age of the primary beneficiary.</li>
        <li><strong>sex:</strong> The gender of the insurance contractor (female or male).</li>
        <li><strong>bmi:</strong> Body Mass Index, an objective measure of body weight based on the ratio of height to weight. The ideal BMI range is 18.5 to 24.9.</li>
        <li><strong>children:</strong> The number of children/dependents covered by the health insurance.</li>
        <li><strong>smoker:</strong> Indicates whether the individual is a smoker.</li>
        <li><strong>region:</strong> The residential area of the beneficiary in the US, categorized into northeast, southeast, southwest, or northwest.</li>
        <li><strong>charges:</strong> The individual medical costs billed by health insurance.</li>
    </ul>
    <p>This dataset is utilized for building predictive models using <strong><em>Decision Tree</em></strong> and <strong><em>Linear Regression</em></strong> to estimate medical costs.</p>

```python
def remove_outliers(df, columns):
  for col in columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 - 1.5 * IQR
    df = df[(df[col] >= lower_bound) & (df[col] <= upper_bound)]
    return df
columns_with_outliers = ['charges']

# remove outliers using IQR
data = remove_outliers(df, columns_with_outliers)
```

<div style="border: 2px solid #2ca8aa; border-radius: 10px; padding: 20px; background-color: #2ca8aa; font-family: 'Algerian', serif; color: #333; text-align: center;">

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler, OneHotEncoder ,StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.compose import make_column_transformer
from sklearn.tree import DecisionTreeRegressor
```

<!-- # **Normalization By Standard Scaling** -->
<div style="border: 2px solid #2ca8aa; border-radius: 10px; padding: 20px; background-color: #2ca8aa; font-family: 'Algerian', serif; color: #333; text-align: center;">
    <h2 style="color: white;">📏 Normalization By Standard Scaling</h2>
</div>

```python
scaler = StandardScaler()
df[['age', 'bmi', 'children']] = scaler.fit_transform(df[['age', 'bmi', 'children']])
```

<!-- # **Baisc Steps** -->
<div style="border: 2px solid #2ca8aa; border-radius: 10px; padding: 20px; background-color: #2ca8aa; font-family: 'Algerian', serif; color: #333; text-align: center;">
    <h2 style="color: white;">Evaluation of Dataset</h2>
</div>

```python
X = df.drop('charges', axis=1)
y = df['charges']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

<!-- # **Baisc Steps** -->
<div style="border: 2px solid #2ca8aa; border-radius: 10px; padding: 20px; background-color: #2ca8aa; font-family: 'Algerian', serif; color: #333; text-align: center;">
    <h2 style="color: white;">🔁 Transformation of Dataset</h2>
</div>

```python
ct = make_column_transformer((MinMaxScaler(), ['age', 'bmi', 'children']), 
                             (OneHotEncoder(handle_unknown = 'ignore'), ['sex', 'smoker', 'region']))
```

```python
# fitting column transformer on training data
ct.fit(X_train)

# transforming training and test data with normalizing (MinMaxScaler) and one hot encodig (OneHotEncoder)
X_train = ct.transform(X_train)
X_test = ct.transform(X_test)
```

<!-- # **Baisc Steps** -->
<div style="border: 2px solid #2ca8aa; border-radius: 10px; padding: 20px; background-color: #2ca8aa; font-family: 'Algerian', serif; color: #333; text-align: center;">
    <h2 style="color: white;">Applying Models (Linear Regression & Decision Tree</h2>
</div>

```python
lr = LinearRegression()
dt = DecisionTreeRegressor(max_depth = 3)

regressors = [('Linear Regression', lr),
               ('Decision Tree', dt)]
```

<!-- # **Baisc Steps** -->
<div style="border: 2px solid #2ca8aa; border-radius: 10px; padding: 20px; background-color: #2ca8aa; font-family: 'Algerian', serif; color: #333; text-align: center;">
    <h1 style="color: white;">COMPARISION</h1>
</div>

```python
for regressor_name, regressor in regressors:
    regressor.fit(X_train, y_train)
    y_pred = regressor.predict(X_test)
    accuracy = round(r2_score(y_test,y_pred),1)*100
    print('{:s} : {:.0f} %'.format(regressor_name, accuracy))
    plt.rcParams["figure.figsize"] = (5,3)
    plt.bar(regressor_name,accuracy)
```
